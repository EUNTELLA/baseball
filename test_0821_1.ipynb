````python
from pathlib import Path

code = r'''
import subprocess
import requests
import time
import threading
from collections import deque

WEBHOOK_URL = "여기에_디스코드_웹훅_URL"

def _send(message):
    try:
        requests.post(
            WEBHOOK_URL,
            json={"content": message},
            timeout=10
        )
    except Exception as e:
        print("[Discord 전송 실패]", e)

def run_discord(command, interval=300, recent_lines=30):
    start = time.time()

    _send(
        "🚀 **Colab 작업 시작**\n"
        f"`{command.splitlines()[0][:150]}`"
    )

    process = subprocess.Popen(
        command,
        shell=True,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1
    )

    recent = deque(maxlen=recent_lines)
    stop_flag = False

    def reporter():
        while not stop_flag:
            time.sleep(interval)

            if process.poll() is not None:
                break

            if recent:
                text = "\n".join(recent)

                if len(text) > 1700:
                    text = text[-1700:]

                _send(
                    "📟 **Colab 최근 로그**\n"
                    f"```text\n{text}\n```"
                )

    thread = threading.Thread(target=reporter, daemon=True)
    thread.start()

    try:
        for line in process.stdout:
            print(line, end="", flush=True)
            recent.append(line.rstrip())

        return_code = process.wait()

    except Exception as e:
        _send(f"🚨 **모니터링 중 오류 발생**\n`{e}`")
        raise

    finally:
        stop_flag = True

    elapsed = time.time() - start

    if recent:
        text = "\n".join(recent)

        if len(text) > 1700:
            text = text[-1700:]

        _send(
            "📟 **마지막 로그**\n"
            f"```text\n{text}\n```"
        )

    if return_code == 0:
        _send(
            "✅ **Colab 작업 완료**\n"
            f"⏱ 실행 시간: {elapsed / 60:.1f}분"
        )
    else:
        _send(
            "🚨 **Colab 실행 오류**\n"
            f"Exit Code: `{return_code}`\n"
            f"⏱ 실행 시간: {elapsed / 60:.1f}분"
        )

    return return_code
'''

path = Path("/content/drive/MyDrive/discord_runner.py")
path.write_text(code, encoding="utf-8")

print("저장 완료:", path)
````
